### Directory Loader

```python
import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)

from langchain_community.document_loaders import DirectoryLoader, TextLoader

# Path to your directory containing files
directory_path = "../data/text_files"

# Set up DirectoryLoader
loader = DirectoryLoader(
    path=directory_path,
    glob="*.txt",                     # Load all .txt files
    loader_cls=TextLoader,            # Use TextLoader for each file
    loader_kwargs={"encoding": "utf-8"} # Pass utf-8 encoding to TextLoader
)

# Load all documents
docs = loader.load()

# Inspect results
print(f"Total documents loaded: {len(docs)}")
for doc in docs:
    print(f"Source: {doc.metadata['source']} | Length: {len(doc.page_content)} characters")
```

### PDF Loader

```python
import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)

from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader,PyMuPDFLoader

# Path where your PDFs are stored
directory_path = "../data/pdf"

loader = DirectoryLoader(
    path=directory_path,
    glob="*.pdf",                  # Use "**/*.pdf" to include subfolders
    loader_cls=PyMuPDFLoader,        # Parses PDF pages
    use_multithreading=True        # Faster loading across multiple files
)

docs = loader.load()
print(docs)
print(f"Total pages/documents loaded: {len(docs)}")
```

```python
from pathlib import Path
import pymupdf4llm
from langchain_core.documents import Document

# Adjust to your folder path
pdf_directory = Path("../data/pdf")  # or Path("../data/pdf_files")
docs = []

for pdf_path in pdf_directory.glob("*.pdf"):
    page_data = pymupdf4llm.to_markdown(str(pdf_path), page_chunks=True)
    
    # Use enumerate to guarantee safe 1-based page numbers
    for page_num, page in enumerate(page_data, start=1):
        # Safely copy existing metadata if available
        meta = page.get("metadata", {}).copy() if isinstance(page.get("metadata"), dict) else {}
        
        # Add source and page safely
        meta["source"] = str(pdf_path)
        meta["page"] = meta.get("page") or meta.get("page_number") or page_num
        
        docs.append(
            Document(
                page_content=page.get("text", ""),
                metadata=meta
            )
        )

print(f"Loaded {len(docs)} pages across all PDFs.")

if docs:
    print("Preview:\n", docs[1].page_content[:300])
    print("\nMetadata:", docs[1].metadata)
```
